**Normalizacion**: Convertir los datos de entrada en un rango de 0 y 1, a este proceso lo llamamos homotecia

**Formula**

$x normalizado = (x - x mínimo ) / (x máximo - x mínimo )$

```python
from sklearn import preprocessing
import numpy as np
```

**MaxMinScale**: Funcion que busca el maximo  el minimo de una funcion

**numpy**: Funcion de python que creada en c que busca, crear ayudar con opreciones como: matrices o vectires, etc

```python
# Carga de datos.
X_train = np.array([[ 1.,-4., 5.],
[ 5., 0., 2.],
[ 0., 6.,-3.]])
print(X_train)
```

**FIT**: Entrena

 **Transform**: Crea otra amtriz remplazando en formula

```python
# Limpieza de datos: normalización.
normalizer = preprocessing.MinMaxScaler()
X_train_norm = normalizer.fit_transform(X_train)
```

```python
# Salida de resultados.
print(X_train_norm)
```

```python
# Ejemplo para ver que los resultados son iguales si aplicamos los métodos␣
#↪"fit" y "transform" por separado
X_train_2 = np.copy(X_train)
# Crea un objeto "fitted_normalizer" llamando al método "fit"
f = normalizer.fit(X_train)
X_train_norm_2 = f.transform(X_train)
# Aplica el objeto sobre los datos "X_train_2" para obtener "X_train_norm_2"
# Comprueba si "X_train_norm" y "X_train_norm_2" son iguales
print(np.array_equiv(X_train_norm, X_train_norm_2))
```

```python
# Normalizar un nuevo conjunto de datos de test (SOLO HAY QUE APLICAR EL MÉTODO␣
#↪"transform", NO EL "fit")
X_test = np.array([[ 1.,-3., 2.],
[ 3., 0., 0.],
[ 0., 4.,-1.]])
# Normalizar los datos de "X_test"
X_test_norm=f.transform(X_test)
```
